### Import the dependencies

```python
from utils import convert_to_json
from metric.evaluator import get_evaluator
import pandas as pd
import nltk
nltk.download('punkt')
```

```python
# Path to News_Generated_In_English.csv archive
file_path = '/content/noticias_generadas_completo_english (1).csv'
df = pd.read_csv(file_path)
```

### Evaluators

```python
task = 'summarization'

# Initialize evaluator for a specific task
evaluator = get_evaluator(task)
```

```python
task = 'dialogue'
# Initialize evaluator for a specific task
evaluator = get_evaluator(task)
```

### LLaMA3 FT Evaluations

```python
# a list of source documents
src_list = (df['Fecha'] + ". " + df['Contenido']).tolist()
# a list of human-annotated reference summaries
ref_list = src_list.copy()  # Asumiendo que ref_list es igual a src_list
# a list of model outputs to be evaluataed
output_list = df['generadaLLAMA3'].tolist()

# Prepare data for pre-trained evaluators
data = convert_to_json(output_list=output_list,
                       src_list=src_list, ref_list=ref_list)
```

```python
# Get multi-dimensional evaluation scores
print("Evaluación LLaMA 3 FT Summarize ")
eval_scores = evaluator.evaluate(data, print_result=True)
```

```python
# Generar las listas necesarias
instruction = "Your task is to write news articles that always include a date, a place, and an event. You cannot invent information that is not given, use formal language."

# Asegurarse de tener al menos 20 elementos para la prueba
sample_size = 20
df_sample = df.sample(n=sample_size, random_state=1) if len(df) >= sample_size else df

src_list = [instruction] * sample_size

# a list of additional context that should be included into the generated response
context_list = (df['Fecha'] + ". " + df['Contenido']).tolist()
# a list of model outputs to be evaluated
output_list = df['generadaLLAMA3'].tolist()

# Prepare data for pre-trained evaluators
data = convert_to_json(output_list=output_list,
                       src_list=src_list, context_list=context_list)
```

```python
# Get multi-dimensional evaluation scores
print("Evaluación LLaMA 3 FT Dialogue")
eval_scores = evaluator.evaluate(data, print_result=True)
```
